 #### Base de dados:

com cerca de 72530 acidentes, nós periodos de 2025-01-01 a 2025-11-02, apresento as seguintes colunas na base de dados

- id = Identificador único do acidente na base de dados. 
- data_inversa = Data da ocorrência do acidente.
- dia_semana = semana ocorrida no acidente 
- horario = Horário da ocorrência do acidente.
- uf = Unidade da Federação (estado) onde ocorreu o acidente.
- br = Número da rodovia federal (BR) onde ocorreu o acidente.
- km = Quilômetro da rodovia onde ocorreu o acidente. 
- municipio = Município da ocorrência do acidente. 
- causa_acidente = Causa presumível do acidente identificada pela PRF. 
- tipo_acidente = Tipo do acidente (colisão frontal, saída de pista, capotamento, atropelamento etc.)
- classificacao_acidente = Classificação da gravidade do acidente (sem vítimas, com vítimas feridas ou com vítimas fatais).
- fase_dia = Período do dia em que ocorreu o acidente (amanhecer, pleno dia, anoitecer, plena noite etc.).
- sentido_via = Sentido da via no local do acidente (crescente ou decrescente). 
- condicao_metereologica = Condição climática no momento do acidente (céu claro, chuva, neblina, vento etc.).

- tipo_pista = Tipo de pista da rodovia (simples, dupla ou múltipla). 
- tracado_via = Característica geométrica da via (reta, curva, cruzamento, viaduto, interseção etc.). 
- uso_solo = Classificação da área onde ocorreu o acidente (urbana ou rural). 
- pessoas = Quantidade total de pessoas envolvidas na ocorrência. 
- mortos = Quantidade de pessoas que morreram em decorrência do acidente. 
- feridos_leves = Quantidade de pessoas com ferimentos leves. 
- feridos_graves = Quantidade de pessoas com ferimentos graves. 
- ilesos = Quantidade de pessoas envolvidas que não sofreram ferimentos. 
- ignorados = Quantidade de pessoas cujo estado físico não foi identificado ou informado. 
- feridos = Total de pessoas feridas (soma de feridos leves e graves). 
- veiculos = Quantidade total de veículos envolvidos no acidente. 
- latitude = Coordenada geográfica de latitude do local do acidente. 
- longitude = Coordenada geográfica de longitude do local do acidente. 
- regional = Superintendência Regional da PRF responsável pela ocorrência. 
- delegacia = Delegacia da PRF responsável pelo atendimento da ocorrência. 
- uop = Unidade Operacional da PRF responsável pelo registro ou atendimento da ocorrência.

#### Bloco 2 - Frequências, rankings e séries simples 

```python
def ranking_municipio(var): 
    
    t = (
        
        prf_acidentes_2025
        .groupby(var) 
        .agg(
            
            acidentes = ("municipio", "count"), 
            acidentes_graves = ("acidente_grave", "sum")
            
        )
        .sort_values("acidentes", ascending=False)
    )
    
    t["pct_acidentes"] = t["acidentes"] / total_acidentes 
    
    return t

ranking_municipio("municipio").head(30)
```

```python
def ranking_causa(var): 
    
    t = (
        
        prf_acidentes_2025
        .groupby(var) 
        .agg(
            
            acidentes = ("causa_acidente", "count"), 
            causas_acidentes = ("acidente_grave", "sum")
            
        )
        .sort_values("acidentes", ascending=False)
    )
    
    t["pct_acidentes"] = t["acidentes"] / total_acidentes 
    
    return t

ranking_causa("causa_acidente").head(30)
```

```python
def ranking_tipos_acidentes(var):
    
    t = (
        
        prf_acidentes_2025
        .groupby(var)
        .agg(
            acidentes = ("tipo_acidente", "count"), 
            causas_acidentes = ("acidente_grave", "sum")
        )
        .sort_values("acidentes", ascending=False)
    )
    
    t["pct_acidentes"] = t["acidentes"] / total_acidentes 
    
    return t

ranking_tipos_acidentes("tipo_acidente")
```

#### Série Mensal 

- #### Criação da coluna "mẽs" no formato Ano-Mês (ex: 2025-01) 

```python
# Agrupa cada acidente no seu respectivo mês, servindo como base para a analise da série temporal 
prf_acidentes_2025["mes"] = (
    
    prf_acidentes_2025["data_inversa"]
    .dt.to_period("M").astype(str)
)
```

- #### Agrupamento dos acidentes por mês 

```python

# Calcula simultaneamente a quantidade total de acidentes, 
# acidentes graves e acidentes em condições climáticas adversas 
serie_mensal = prf_acidentes_2025.groupby("mes").agg(
    
    acidentes = ("municipio", "count"), 
    acidentes_graves = ("acidente_grave", "sum"), 
    clima_adverso = ("clima_adverso", "sum")
)
```

- #### Cálculo do percentual mensal de acidentes graves 

```python
# Permite observar se a proporção de acidentes graves 
# oscila ao longo dos meses 
serie_mensal["pct_graves"] = (
    
    serie_mensal["acidentes_graves"] / 
    serie_mensal["acidentes"]
)
```

- #### Taxa de mortalidade 

```python
serie_mensal["taxa_mortalidade"] = (
    
    prf_acidentes_2025.groupby("mes")["mortos"].sum() / serie_mensal["acidentes"]
)
```

- #### ìndice de gravidade 

```python
serie_mensal["indice_gravidade"] = (
    
    prf_acidentes_2025.groupby("mes").apply(
        
        lambda prf_acidentes_2025: (prf_acidentes_2025["mortos"].sum()*3 +
                                    prf_acidentes_2025["feridos_graves"].sum() * 2
                                   + prf_acidentes_2025["feridos_leves"].sum() * 1)
    ) / serie_mensal["acidentes"]
)
```

Além disso, a taxa de mortalidade e o índice de gravidade são indicadores gerais e não estabelecem, isoladamente, uma relação de causa e efeito.  - Leitura: A série mensal permite acompanhar a distribuição dos acidentes ao longo de 2025 e comparar a quantidade total de ocorrências com os acidentes graves. ### - Análise bivariada focada no alvo 


#### Este é o coração da EDA. 

- Comparamos cada variável explicativa com acidentes_problematicos, sempre relatando volume e proporção - nunca só uma das duas 

```python
#### Criação de uma variável-alvo binária: acidente com vítimas fatais (1) ou não (0) 
prf_acidentes_2025["acidente_fatal"] = prf_acidentes_2025["classificacao_acidente"].fillna("").apply(
    lambda x: 1 if any(p in x.lower() for p in ["fatais", "fatal", "óbito", "morte"]) else 0
)
```

- Definição da função de análise bivariada adaptada para a base da PRF
*** var: variável explicativa (ex: "uf", "causa_acidente" etc.) 
*** min_acidentes: mínimo de acidentes para inclusão na análise (evita categorias irrelevantes) 

```python
def analise_bivariada_prf(prf_acidentes_2025, var, min_acidentes=1): 
    
    tab = prf_acidentes_2025.groupby(var).agg(
        
        # total de acidentes na categoria
        acidentes_total=("id", "count"), 
        
        # total de acidentes com vítimas fatais na categoria
        acidentes_fatais=("acidente_fatal", "sum"), 
        
        # total de mortes na categoria 
        mortos=("mortos", "sum"), 
        
        # total de feridos na categoria
        feridos=("feridos", "sum")
    )
    
    ### proporção condicional de acidente fatal
    tab["pct_acidente_fatal"] = tab["acidentes_fatais"] / tab["acidentes_total"]  
    
    ### filtro opcional de volume mínimo 
    tab = tab[tab["acidentes_total"] >= min_acidentes] 
    
    return tab.sort_values("pct_acidente_fatal", ascending = False) 
```

```python
#### Exemplo de uso com algumas variáveis explicativas 
variaveis_explicativas = ["uf", "causa_acidente", "fase_dia", "condicao_metereologica"] 
```

#### Achados observados na base PRF 2025 

A análise bivariada permitiu identificar diferenças relevantes na proporção de acidentes fatais conforme a UF, 
causa do acidente, condição meteorológica, tipo de acidente e uso do solo. A taxa global de referência é de 
aproximadamente 7,2% de acidentes fatais 

Entre os estados, Maranhão (19%) e Pará (17%) apresentaram as maiores proporções de acidentes fatais. 
Em relação às causas dos acidentes, destacam-se "Pedestre andava na pista" (41%) e "Transitar na contramão" (30%), ambas muito superiores à taxa global. Nos tipos de acidente, Atropelamento de Pedestre (30%), e Colisão 
Frontal(29%) também apresentaram proporções significativamente superiores à referência. 


Na dimensão meteorológica, acidentes ocorridos sob nevoeiro / neblina (11%) apresentam proporção de fatalidade 
acima da média geral. Já quanto ao uso do solo, os acidentes classificados como ocorridos em área de "Não" 
uso do solo apresentaram 9%, também acima da taxa global. 

Esses resultados representam associações observadas nos dados, não permitindo afirmar, isoladamente, que uma 
determinada condição seja a causa dos acidentes fatais. Além disso, algumas categorias dos quadros de município, KM e traçado da via apresentaram taxas de 100%, porém com apenas um ou poucos acidentes registrados. Por isso, essas categorias devem ser interpretadas com cautela, pois o pequeno número de observações pode produzir percentuais elevados sem representar um padrão geral da população.
